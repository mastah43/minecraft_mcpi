```python
from mcpi.minecraft import Minecraft
from mcpi.block import Block

# mcpi is supported on minecraft java servers 
# via the plugin https://www.curseforge.com/minecraft/bukkit-plugins/raspberryjuice

mc = Minecraft.create(address = "localhost", port = 4711)
mc.postToChat("connected to minecraft java server")
```

```python
print("player position:", mc.player.getPos())
```

```python
from math import cos, sin, pi
from time import sleep
from mcpi.event import ChatEvent
from mcpi.block import WOOD
from mcpi.block import SANDSTONE
from mcpi.block import STONE
from mcpi.block import WOOD_PLANKS
from mcpi.block import WOOL
from mcpi.block import AIR

def build_pyramid(x, y, z):
    print(f"building pyramid: x={x}, y={y}, z={z}")
    height = 5
    for h in range(height):
        for dx in range(-height + h, height - h + 1):
            for dz in range(-height + h, height - h + 1):
                mc.setBlock(x + dx, y + h, z + dz, SANDSTONE.id)

def build_arc(x, y, z):
    print(f"building arc: x={x}, y={y}, z={z}")
    radius = 10
    for angle in range(0, 180, 5):
        rad = angle * pi / 180
        dx = int(radius * cos(rad))
        dy = int(radius * sin(rad))
        dz = 0
        mc.setBlock(x + dx, y + dy, z + dz, STONE.id)

def build_house(x, y, z):
    def build_base(x, y, z, base_size):
        mc.setBlocks(x - base_size, y, z - base_size,
                    x + base_size, y, z + base_size, WOOL.id)
        
    def build_walls(x, y, z, base_size, wall_height):
        wall_block_id = WOOD_PLANKS.id
        mc.setBlocks(x - base_size, y, z - base_size,
                    x - base_size, y + wall_height, z + base_size, wall_block_id)
        mc.setBlocks(x + base_size, y, z - base_size,
                    x + base_size, y + wall_height, z + base_size, wall_block_id)
        mc.setBlocks(x - base_size, y, z - base_size,
                    x + base_size, y + wall_height, z - base_size, wall_block_id)
        mc.setBlocks(x - base_size, y, z + base_size,
                    x + base_size, y + wall_height, z + base_size, wall_block_id)
        
    def build_door(x, y, z):
        door_height = 2
        door_width = 2
        mc.setBlocks(x, y, z,
                    x, y + door_height - 1, z + door_width - 1, AIR.id)


    def build_roof(x, y, z, base_size):
        roof_height = 3
        roof_block_id = WOOD.id
        for dy in range(0, roof_height):
            mc.setBlocks(
                x - base_size + dy, y + dy, z - base_size + dy, 
                x + base_size - dy, y + dy, z + base_size - dy, 
                roof_block_id)
            if dy < roof_height - 1:
                mc.setBlocks(
                    x - base_size + dy + 1, y + dy, z - base_size + dy + 1, 
                    x + base_size - dy - 1, y + dy, z + base_size - dy - 1, 
                    AIR.id)

    print(f"building house: x={x}, y={y}, z={z}")
    base_size = 5
    wall_height = 4
    build_base(x, y-1, z, base_size)
    build_walls(x, y, z, base_size, wall_height)
    build_door(x - base_size, y, z)
    build_roof(x, y + wall_height, z, base_size)


def listenForChatMessages():
    while True:
        chatEvents = mc.events.pollChatPosts()
        for chatEvent in chatEvents:
            msg = chatEvent.message
            pos = mc.entity.getPos(chatEvent.entityId)
            x, y, z = int(pos.x), int(pos.y), int(pos.z)
            print("chat message: ", msg)
            if msg == "pyramid":
                build_pyramid(x,y,y)
            elif msg == "arc":
                build_arc(x,y,z)
            elif msg == "house":
                build_house(x,y,z)
        sleep(1)

listenForChatMessages()
```
